```python
# method 2
def map_vechicles(v):
    if any(x in v for x in ['Car','Taxi/Private hire car']):
        return 'Cars'
    elif any(x in v for x in ['Motorcycle over 500cc','Motorcycle 125cc and under','Motorcycle 50cc and under','Motorcycle over 125cc and up to 500cc']):
        return 'Bike'
    elif any(x in v for x in ['Van / Goods 3.5 tonnes mgw or under','Bus or coach (17 or more pass seats)','Goods 7.5 tonnes mgw and over','Goods 7.5 tonnes mgw and over','Goods over 3.5t. and under 7.5t']):
        return 'Van'
    elif any(x in v for x in ['Bus or coach (17 or more pass seats)','Minibus (8 - 16 passenger seats)']):
        return 'Bus'
    elif any(x in v for x in ['Agricultural vehicle']):
        return 'Agricultural vehicle'
    else:
        return 'Other vechicle'
```

```python
file['vehicle_category']=file['vehicle_type'].map(map_vechicles)
file
```

```python
acc_by_vehicle=(
    file.groupby('vehicle_category')['number_of_casualties']
    .sum()
    .sort_values(ascending=False)
    .to_frame()
)
acc_by_vehicle
```

```python
# Area – wise Percentage(%) and Total Number of Casualties
area_wise_count=(
    file.groupby('urban_or_rural_area')['number_of_casualties'].sum()
    .sort_values(ascending=False)
    .to_frame(name='count')
)
area_wise_percent=(
    file.groupby('urban_or_rural_area')['number_of_casualties'].sum()/file['number_of_casualties'].sum()*100
).to_frame(name='percent')
```

```python
df1=pd.concat([area_wise_count,area_wise_percent],axis=1)
df1
```

```python
def map_light(v):
    if any(x in v for x in ['Darkness - lights lit','Darkness - no lighting','Darkness - lighting unknown','Darkness - lights unlit']):
        return 'Night'
    else :
        return 'Day'
```

```python
file['light_type']=file['light_conditions'].map(map_light)
file
```

```python
# Count of Casualties By Light Conditions
casualties_by_light=(
    file.groupby('light_type')['number_of_casualties']
    .sum()
    .sort_values(ascending=False)
    .to_frame(name='count')
)
casualties_by_light
```

```python
# 14. Percentage (%) and Segregation of Casualties by Different Light Conditions
casualties_percent=(
    file.groupby('light_type')['number_of_casualties']
    .sum()/file['number_of_casualties'].sum()*100
).to_frame(name='percentage')
df2=pd.concat([casualties_by_light,casualties_percent],axis=1)
df2
```
